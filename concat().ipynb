ROW-WISE CONCATENATION.(it's Default Concatenation)

```python
df1 = pd.DataFrame({
    'pid': ['p1', 'p2', 'p3'],
    'pname': ['TV', 'Laptop', 'AC'],
    'price': [20000, 100000, 65000]})

# 2. Create the second DataFrame with the SAME column names
df2 = pd.DataFrame({
    'pid': ['p4', 'p5'],
    'pname': ['Mobile', 'Watch'],
    'price': [30000, 15000]})

# 3. Concatenate row-wise (axis=0)
df3= pd.concat([df1, df2], ignore_index=True)
print(df3)
```

merge() -> used to fetch data from multiple dATAframes.

syntax: pd.merge(df1,df2,on="common_clmn",how='type_of_join')

```python
r=pd.merge(df1,df2,how="cross")
print(r)
```

```python
r=pd.merge(df1,df2,on="pid",how="left")
print(r)
```

```python
r=pd.merge(df1,df2,on="pid",how="right")
print(r)
```

```python
r=pd.merge(df1,df2,on="pid",how="outer")
print(r)
```

```python
r=pd.merge(df1,df2,on="pid",how="inner")
print(r)
```
